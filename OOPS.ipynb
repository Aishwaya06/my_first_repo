```python
class Car:
  def __init__(self,name,color,model,price):
    self.name = name
    self.color = color
    self.model = model
    self.price = price

  def display(self):
    print("My Car is ", self.name,"of",self.color,"color")
    print("It is of",self.model,"model of price ",self.price)

car1 = Car("Thar","Black",2026,15000000) #obj1
car2 = Car("Fortuner","White",2022,20000000) #obj2

car1.display()
car2.display()

# self is not mandatory , but it is a standard parameter

```

```python
# What if i want to acces private data member outside the class

# Getter - Getting/accessing the private data values
# Setter - setting/updating the private data values
```

```python
class person:
  def __init__(self,name,job):
    self.name = name # public
    self.__job = job # private

  def getter(self):
    print("The job of the person is: ",self.__job)

ob = person("Sriram","Data Analyst") # object
print(ob.name)
ob.getter()
```

```python
# Getter
# Setter

class person:
  def __init__(self,name,job):
    self.name = name # public
    self.__job = job # private

  def getter(self):
    print("The job is : ",self.__job)

  def setter(self,newjob):
    self.__job = newjob

ob1 = person("Sriram","Data Analyst")
print(ob1.name)
ob1.setter("Data Scientist")
ob1.getter()

```

```python
# Inheritance

# Where a class/classes inherits the properties of its parents
# Parent Class - The class whose properties are going to be inherited
# Child Class - The class that inherits the properties of the parent class
```

```python
# Code Reusability
```

```python
# Types of Inheritance

# Single = a -> b(One parent -> One child)
# Multiple = a,b,c,d, -> e (Multiple Parent -> One child)
# Multilevel = a -> b -> c -> d -> e (One Person at multiple level)
# Heirarchical = a -> b,c,d... (Single Parent -> Multiple Children)
# Hybrid = Combining diff types of inheritance
```

```python
# Multilevel - CEO -> VP -> AVP -> Manager
```

```python
class Employee: # Parent class
  def __init__(self,name,empid):
    self.name = name
    self.empid = empid

  def disp(self):
    print("The name of the employee is ",self.name,"whose employee id is ",self.empid)

class Tech(Employee): # child class
  def __init__(self,name,empid,team):
    #super().__init__(name,empid) # super function
    Employee.__init__(self,name,empid) # Direct acces
    self.team = team

  def display(self):
    print(self.name,self.empid,self.team)

ob = Tech("Preethi",9789,"Python") # object
ob.display()

# using the object of the child class , i can access the parent class properties - Inheritance


```

```python
# Multiple Inheritance

class mother:
  def __init__(self,mname):
    self.mname =mname

class father:
  def __init__(self,fname):
    self.fname = fname

class child(mother,father):
  def __init__(self,mname,fname,cname):
    mother.__init__(self,mname)
    father.__init__(self,fname)
    self.cname = cname

  def display(self):
    print(self.mname,self.fname,self.cname)

ob2 = child("Lilly Potter","James Potter","Harry Potter")
ob2.display()


```
