```python
class Employee: # Parent class
  def __init__(self,name,empid):
    self.name = name
    self.empid = empid

  def disp(self):
    print("The name of the employee is ",self.name,"whose employee id is ",self.empid)

class Tech(Employee): # child class
  def __init__(self,name,empid,team):
    #super().__init__(name,empid) # super function
    Employee.__init__(self,name,empid) # Direct acces
    self.team = team

  def display(self):
    print(self.name,self.empid,self.team)

ob = Tech("Preethi",9789,"Python") # object
ob.display()

# using the object of the child class , i can access the parent class properties - Inheritance


```

```python
# Multiple Inheritance

class mother:
  def __init__(self,mname):
    self.mname =mname

class father:
  def __init__(self,fname):
    self.fname = fname

class child(mother,father):
  def __init__(self,mname,fname,cname):
    mother.__init__(self,mname)
    father.__init__(self,fname)
    self.cname = cname

  def display(self):
    print(self.mname,self.fname,self.cname)

ob2 = child("Lilly Potter","James Potter","Harry Potter")
ob2.display()


```

```python
# Create a movie Booking System

# Class named as MovieTheatre(movie,seats)

# Methods:
# Book_ticket(count) - if count < seats - u can book , otherwise u cannot

# cancel_ticket(count)
# available_seats()

# Time till 9:28 pm
```

```python
class MovieTheatre:

    def __init__(self, movie, seats):
        self.movie = movie
        self.seats = seats

    def book_ticket(self, count): # avai- 10 , count - 2 , avai= 10-2 = 8
        if count <= self.seats:
            self.seats = self.seats - count
            print(count, "tickets booked successfully")
        else:
            print("Cannot book tickets. Not enough seats available")

    def cancel_ticket(self, count): #
        self.seats = self.seats + count
        print(count, "tickets cancelled successfully")

    def available_seats(self):
        print("Available seats:", self.seats)


theatre = MovieTheatre("Avengers", 100) # object

theatre.available_seats()

theatre.book_ticket(20)
theatre.available_seats()

theatre.cancel_ticket(5)
theatre.available_seats()
```

```python
# Create a Student grade system

# Create a class named as Student

def __init__(self):
  self.scores = []

def add_score(self,score):
  self.scores.append(score)

# It should have method:
# add_score()
# calculate_avg()
# display_grade()

# 90+ = O
# 80+ = A
# 70+ = B
# 50+ = C
# else = Fail

# Time till 9:58 pm
```

```python
# 80 , 89, 65,34 = sum/len(list)

# Based upon the avg - give the grade
```

```python
class Student:

    def __init__(self):
        self.scores = []

    def add_score(self, score):
        self.scores.append(score)

    def calculate_avg(self):
        return sum(self.scores) / len(self.scores)

    def display_grade(self):
        avg = self.calculate_avg()

        if avg >= 90:
            print("Grade: O")
        elif avg >= 80:
            print("Grade: A")
        elif avg >= 70:
            print("Grade: B")
        elif avg >= 50:
            print("Grade: C")
        else:
            print("Grade: Fail")


# Create object
student = Student()

# Add scores
student.add_score(90)
student.add_score(80)
student.add_score(70)
student.add_score(60)

# Display average and grade
print("Average:", student.calculate_avg())
student.display_grade()
```

```python
# Create a Employee salary System

# parent class - Employee, method - calculate_bonus()
# child classes - Manager , Team_Lead

# Both classes should override calculate_bonus method
# A manager gets a bonus of 20% of their salary
# A Team lead gets a bonus of 10% of their salary
```

```python
class Employee:
    def __init__(self, salary):
        self.salary = salary
    def calculate_bonus(self):
        return self.salary * 0.05

class Manager(Employee):
    def calculate_bonus(self):
        return self.salary * 0.20

class TeamLead(Employee):
    def calculate_bonus(self):
        return self.salary * 0.10

manager = Manager(200000)
team_lead = TeamLead(150000)
print("Manager bonus:", manager.calculate_bonus())
print("Team Lead bonus:", team_lead.calculate_bonus())
```

```python
from abc import ABC , abstractmethod

class Vehicle(ABC): # Abstract Class
  @abstractmethod
  def start(self):
    pass

  @abstractmethod
  def stop(self):
    pass

class Car(Vehicle): # Concrete class implements the methods of abstract class
  def start(self):
    print("The Car started")
  def stop(self):
    print("The Car stopped")

ob = Car()
ob.start()
```

```python
# Practise Question

# Design a Ride Booking System

# It should have an abstract class named RideBooking
# Having an abstract method named calculate_fare()
# Create two concrete classes BikeRide and CarRide
# These should implement the abstract class method differently
# use python's abc module
```

```python
from abc import ABC, abstractmethod

class RideBooking(ABC):
  @abstractmethod
  def calculate_fare(self):
    pass

class BikeRide(RideBooking):
  def calculate_fare(self):
    return 100

class CarRide(RideBooking):
  def calculate_fare(self):
    return 500


bike = BikeRide()
car = CarRide()

print("BikeRide fare:",bike.calculate_fare())
print("CarRide fare:",car.calculate_fare())
```

```python
# Create an Ice Cream Shop

# Create a class named IceCream(flavour,stock)
# Methods - serve(), restock(), checkstock()

# Time till 9:54 pm
```

```python
class IceCream:
  def __init__(self,flavour,stock):
    self.flavour=flavour
    self.stock=stock
  def serve(self,quantity):
    if quantity<=self.stock:
      self.stock-=quantity
      print("IceCream is served")
    else:
      print("Icecream cannot be served")
  def restock(self,restock_quantity):
    self.stock+=restock_quantity
  def checkstock(self):
    print("IceCream flavour is "+self.flavour)
    print("Stock present is",self.stock)
ob1=IceCream("Butterscotch",1000)
ob1.serve(300)
ob1.restock(400)
ob1.checkstock()
```

```python
# Create an abstract class named Shape
# Child classes - Circle(self,radius), square(self,side)
# Both these classes should implement area()

# Time till 9:23 pm
```

```python
from abc import ABC, abstractmethod
import math


class Shape(ABC):
    @abstractmethod
    def area(self):
        pass


class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * self.radius * self.radius


class Square(Shape):
    def __init__(self, side):
        self.side = side

    def area(self):
        return self.side * self.side


c = Circle(5)
print("Area of Circle:", c.area())

s = Square(4)
print("Area of Square:", s.area())
```
