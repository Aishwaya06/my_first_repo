```python
def check_vowels_name(name):
    count = 0
    for i in name:
        if i in "aeiouAEIOU":
            count += 1
    print(count)

check_vowels_name("AnIsHa")
```

```python
def count_vowels(s):
    vowels = "aeiouAEIOU"
    count = 0
    for i in s:
        if i in vowels:
            count += 1
    return count


text = "Hello World"
print(count_vowels(text))
```

```python
def check_prime(num):
    if num < 2:
        print("Not a prime number")
    else:
        for i in range(2, num):
            if num % i == 0:
                print("Not a prime number")
                break
        else:
            print("Prime number")

check_prime(5)
```

```python
# How to handle the Exceptions

# try - write the code that might cause an exception
# except - Handles the exception if it occurs
# finally - This block will execute whether an exception occurs or not
```

```python
try:
  print(10/0)
except:
  print("Division by zero is not possible")
finally:
  print("End of the program")

```
